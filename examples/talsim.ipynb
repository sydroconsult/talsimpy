## Manipulating a dataset

A dataset can be manipulated by setting simulation options and calibration parameters

```python
tds = talsim.TalsimDataset(r"data/test_kal-faktor", "Test_KAL-Faktor")
```

```python
# setting simulation options (ALL file)
options = {
    "SimStart": datetime.datetime(2020, 11, 2),
    "SimEnd": datetime.datetime(2021, 7, 30),
    "ResultForStates": "",
    "NonexistentOption": 574,
}

tds.set_sim_options(options)
```

```python
# set calibration parameters (KAL file)
params = {
    "RetKonFakNat": 2,
    "RetKonFakInt": 1,
    "NFK": 1.4,
    "MAXINF": 10,
    "NonexistentParameter": "test",
}

tds.set_calibration_parameters(params)
```

More possibilities for manipulating a dataset are provided by a template system:

* Template files must be named like the original file but with an additional extension "template", e.g. `dataset.EZG.template` and will replace the original file after template processing.
* Template files can contain placeholders for variables enclosed with curly braces, e.g. `{variable_name}`
* Placeholders can optionally contain a format specifier according to https://docs.python.org/3/library/string.html#format-specification-mini-language, examples:
  * `{variable_name:8.3f}` for a float with max length 8 and up to 3 decimal places
  * `{variable_name:>8}` for an integer or string, right-aligned and filled (or cut) to a total length of 8 characters
  * `{variable_name:%d.%m.%Y %H:%m}` for a datetime object
* The method `TalsimDataset.process_templates()` will process all the template files in a dataset by replacing placholders with values

```python
# process template files
vars = {
    "variable_name": "value",
    "other_variable": 12345,
}
tds.process_templates(vars)
```
